# 📚 Kütüphane Yönetim Sistemi

Tkinter ile hazırlanmış grafik arayüzlü uygulama.

```

import tkinter as tk
from tkinter import messagebox

# ------------------ Kitap Sınıfı ------------------
class Kitap:
    def __init__(self, ad, yazar, yil):
        self.ad = ad
        self.yazar = yazar
        self.yil = yil

    def __str__(self):
        return f"{self.ad} | {self.yazar} | {self.yil}"


# ------------------ Kütüphane Sınıfı ------------------
class Kutuphane:
    def __init__(self):
        self.kitaplar = []

    def kitap_ekle(self, kitap):
        self.kitaplar.append(kitap)

    def kitap_sil(self, ad):
        for kitap in self.kitaplar:
            if kitap.ad.lower() == ad.lower():
                self.kitaplar.remove(kitap)
                return True
        return False

    def isme_gore_ara(self, ad):
        return [k for k in self.kitaplar if ad.lower() in k.ad.lower()]

    def yazara_gore_ara(self, yazar):
        return [k for k in self.kitaplar if yazar.lower() in k.yazar.lower()]


# ------------------ Arayüz (GUI) ------------------
class Uygulama:
    def __init__(self, pencere):
        self.kutuphane = Kutuphane()

        pencere.title("Kütüphane Yönetim Sistemi")
        pencere.geometry("700x550")
        pencere.configure(bg="#e6e6e6")

        self.baslik_font = ("Segoe UI", 18, "bold")
        self.label_font = ("Segoe UI", 11, "bold")
        self.entry_font = ("Segoe UI", 11)
        self.buton_font = ("Segoe UI", 11, "bold")
        self.liste_font = ("Consolas", 11)

        tk.Label(
            pencere,
            text="📚 Kütüphane Yönetim Sistemi",
            font=self.baslik_font,
            bg="#e6e6e6",
            fg="#2c3e50"
        ).grid(row=0, column=0, columnspan=2, pady=15)

        tk.Label(pencere, text="Kitap Adı", font=self.label_font, bg="#e6e6e6")            .grid(row=1, column=0, sticky="e", padx=10, pady=5)
        self.entry_ad = tk.Entry(pencere, font=self.entry_font, width=30)
        self.entry_ad.grid(row=1, column=1, pady=5)

        tk.Label(pencere, text="Yazar", font=self.label_font, bg="#e6e6e6")            .grid(row=2, column=0, sticky="e", padx=10, pady=5)
        self.entry_yazar = tk.Entry(pencere, font=self.entry_font, width=30)
        self.entry_yazar.grid(row=2, column=1, pady=5)

        tk.Label(pencere, text="Yayın Yılı", font=self.label_font, bg="#e6e6e6")            .grid(row=3, column=0, sticky="e", padx=10, pady=5)
        self.entry_yil = tk.Entry(pencere, font=self.entry_font, width=30)
        self.entry_yil.grid(row=3, column=1, pady=5)

        butonlar = [
            ("Kitap Ekle", self.kitap_ekle),
            ("Kitap Sil", self.kitap_sil),
            ("İsme Göre Ara", self.isme_gore_ara),
            ("Yazara Göre Ara", self.yazara_gore_ara),
            ("Tüm Kitapları Listele", self.tum_kitaplari_listele),
        ]

        for i, (text, cmd) in enumerate(butonlar):
            tk.Button(
                pencere,
                text=text,
                font=self.buton_font,
                width=25,
                bg="#3498db",
                fg="white",
                activebackground="#2980b9",
                command=cmd
            ).grid(row=4+i, column=0, columnspan=2, pady=4)

        self.liste = tk.Listbox(
            pencere,
            width=80,
            height=10,
            font=self.liste_font
        )
        self.liste.grid(row=9, column=0, columnspan=2, pady=15)

    def kitap_ekle(self):
        ad = self.entry_ad.get()
        yazar = self.entry_yazar.get()
        yil = self.entry_yil.get()

        if ad and yazar and yil:
            self.kutuphane.kitap_ekle(Kitap(ad, yazar, yil))
            messagebox.showinfo("Başarılı", "Kitap eklendi!")
            self.temizle()
        else:
            messagebox.showwarning("Hata", "Tüm alanları doldurun!")

    def kitap_sil(self):
        if self.kutuphane.kitap_sil(self.entry_ad.get()):
            messagebox.showinfo("Başarılı", "Kitap silindi!")
        else:
            messagebox.showerror("Hata", "Kitap bulunamadı!")

    def isme_gore_ara(self):
        self.liste.delete(0, tk.END)
        for kitap in self.kutuphane.isme_gore_ara(self.entry_ad.get()):
            self.liste.insert(tk.END, kitap)

    def yazara_gore_ara(self):
        self.liste.delete(0, tk.END)
        for kitap in self.kutuphane.yazara_gore_ara(self.entry_yazar.get()):
            self.liste.insert(tk.END, kitap)

    def tum_kitaplari_listele(self):
        self.liste.delete(0, tk.END)
        for kitap in self.kutuphane.kitaplar:
            self.liste.insert(tk.END, kitap)

    def temizle(self):
        self.entry_ad.delete(0, tk.END)
        self.entry_yazar.delete(0, tk.END)
        self.entry_yil.delete(0, tk.END)


pencere = tk.Tk()
Uygulama(pencere)
pencere.mainloop()
```
